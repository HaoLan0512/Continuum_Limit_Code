```python
import numpy as np
import pandas as pd
from math import pi

def _fix_global_phase(psi: np.ndarray):
    """Rotate each eigen‑vector so its first non‑zero entry is real‑positive."""
    out = psi.copy()
    for n in range(out.shape[1]):          # loop over columns   ──┐
        for c in out[:, n]:               # scan entries within  │
            if abs(c) > 1e-14:            # pivot found         │
                out[:, n] /= c/abs(c)     # remove that phase   │
                break                     # next eigen‑vector   └─┐
    return out

# --- make a 3×3 matrix of eigen‑vectors, each with a random global phase ----
base_cols = [
    np.array([1+0j, 2j, -1+0j]),          # some arbitrary shapes
    np.array([0+0j, 1-1j, 0+0.5j]),
    np.array([0+0j, 0+0j, 1+0j]),
]
phases = [np.exp(1j*pi/3), np.exp(1j*1.2), np.exp(-1j*0.6)]
psi     = np.column_stack([p*v for p, v in zip(phases, base_cols)])

def show(A, title):
    print(title)
    display(pd.DataFrame(A, columns=[f"vec{n}" for n in range(A.shape[1])]
                         ).applymap(lambda z: f"{z.real:+.3f}{z.imag:+.3f}j"))
    
show(psi, "Original eigen‑vectors with arbitrary phases")

# --- apply the routine ------------------------------------------------------
psi_fixed = _fix_global_phase(psi)
show(psi_fixed, "After _fix_global_phase()")
```
